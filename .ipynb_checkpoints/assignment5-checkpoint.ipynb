Option A: Gutenberg - 'austen-emma.txt'

```python
import numpy as np
import matplotlib.pyplot as plt
import nltk
from nltk.corpus import stopwords
from nltk.corpus import gutenberg
from nltk.tokenize import word_tokenize

from gensim.models import Word2Vec, FastText
```

# Part A  - Text Preprocessing

## A1 Load the corpus

```python
nltk.download('gutenberg')
```

```python
emma = gutenberg.raw('austen-emma.txt')
print(f"Total Number of Character: {len(emma)}")
```

```python
print(f"Total Number of Tokens: {len(gutenberg.words('austen-emma.txt'))}")
```

## A2 Preprocess

In the preprocessing, stop words are removed since they have little meaning to the sentence. Stemming and lemmatization are not used as the they sometimes slightly change the meaning of a word.

```python
nltk.download("stopwords")
stop_words = set(stopwords.words("english"))
```

```python
def preprocess(text: str):
    tokens = word_tokenize(text.lower())
    tokens = [token for token in tokens if token.isalpha() and token not in stop_words]
    return tokens

corpus = preprocess(emma)
```

```python
print(f"Total Number of Tokens after preprocessing: {len(corpus)}")
```

```python
print(f"Vocabulary size: {len(list(set(corpus)))}")
```

```python
from collections import Counter

top_20 = Counter(corpus).most_common(20)
top_20
```

vectorization: minimal preprocessing to avoid removing the original context  
language modeling: better to preserve punctuation and casing as they provide meaning to the sentence  
embeddings: remove meaningless word e.g. stop word and ensure that meaningful distinctions are preserved

# Part B - Text Representation

## B1. Create documents

Fixed-sized chunking is used to balance the document size.

```python
def chunk_text(tokens, chunk_size):
    docs = []
    for i in range(0, len(tokens), chunk_size):
        chunk = tokens[i:i + chunk_size]
        docs.append(' '.join(chunk))
    return docs

tokens = nltk.word_tokenize(emma)
chunk_size = 750
docs = chunk_text(tokens, chunk_size)
len(docs)
```

## B2. Vectorize

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
count_vec = CountVectorizer(stop_words='english')
bow_matrix = count_vec.fit_transform(docs)

tfidf_vec = TfidfVectorizer(stop_words='english')
tfidf_matrix = tfidf_vec.fit_transform(docs)

print("Shape of Bag-of-Words matrix:", bow_matrix.shape)
print("Shape of TF-IDF matrix:", tfidf_matrix.shape)
```

```python
def top_tfidf_terms_for_doc(doc_index, top_k=15):
    feature_array = np.array(tfidf_vec.get_feature_names_out())
    tfidf_sorting = np.argsort(tfidf_matrix[doc_index].toarray()).flatten()[::-1]
    top_terms = feature_array[tfidf_sorting][:top_k]
    top_scores = tfidf_matrix[doc_index].toarray().flatten()[tfidf_sorting][:top_k]
    print(f"\nTop {top_k} TF-IDF terms for document {doc_index}:")
    for term, score in zip(top_terms, top_scores):
        print(f"{term:20s} {score:.4f}")
    return top_terms

top_tfidf_terms_for_doc(0)
top_tfidf_terms_for_doc(10)
```

These two documents have different content so their results are different. Document 0 is likely to be related to relationships while document 10 has content related to egg.

```python
from sklearn.metrics.pairwise import cosine_similarity
cosine_sim = cosine_similarity(tfidf_matrix)
np.fill_diagonal(cosine_sim, 0)
max_sim_idx = np.unravel_index(np.argmax(cosine_sim), cosine_sim.shape)
most_similar_score = cosine_sim[max_sim_idx]

print(f"\nMost similar document pair: {max_sim_idx}, similarity = {most_similar_score:.4f}")

surprise_pair = (0, 2)
print(f"Surprising pair: {surprise_pair} Score: {cosine_sim[surprise_pair]:.4f}")

sim_subset = cosine_sim[:5, :5]
print("\nCosine similarity matrix (first 5 documents):")
print(np.round(sim_subset, 3))
```

```python
# Although Doc 0 and 2 are close to each other with similar content, they have a low similarity score. 
# It may be due to the fact that document 2 includes more rare words relatively in the whole corpus.
top_tfidf_terms_for_doc(0)
top_tfidf_terms_for_doc(2)
```

# Part C - Word Embeddings

## C1. Prepare training data

```python
emma_sents = gutenberg.sents('austen-emma.txt')

def preprocess_sents(tokens: list):
    tokens_lower = [token.lower() for token in tokens]
    tokens_new = [token for token in tokens_lower if token.isalpha()]
    return tokens_new

docs = [preprocess_sents(sent) for sent in emma_sents]
docs[0]
```

## C2. Train Word2Vec

```python
w2v = Word2Vec(
    sentences=docs,
    vector_size=100,
    window=5,
    min_count=3,
    sg=0,
    epochs=10
)
```

## C3. Explore similarity

```python
targets = ["emma", "every", "weston", "one", "harriet"]
for target in targets:
    print(f"\nMost similar words to '{target}':")
    for word, score in w2v.wv.most_similar(target, topn=10):
        print(f"{word:15s} {score:.3f}")
```

The neighbours make sense to me as the names like emma are matched to the other names.

## C4. Analogies (vector arithmetic)

```python
# Query 1 (Expected: influence)
w2v.wv.most_similar(
    positive=["rich", "power"],
    negative=["wealth"],
    topn=10
)
```

```python
# Query 2 (Expected: girl)
w2v.wv.most_similar(
    positive=["man", "boy"],
    negative=["woman"],
    topn=10
)
```

```python
# Query 3 (Expected: rich)
w2v.wv.most_similar(
    positive=["many", "poor"],
    negative=["little"],
    topn=10
)
```

The results are not accurate and it is mainly due to the domain of the corpus being a story and it tends to use more complicated words instead of simple words.
